```python
import pandas as pd

movies = pd.read_csv("Movies_clean(2).csv")
activity = pd.read_csv("Activity_User(2).csv")

print("=" * 60)
print("DATA PROFILING REPORT")
print("=" * 60)

print("\n1. DATA TYPE VALIDATION")

movies["release_year_num"] = pd.to_numeric(
    movies["release_year"], errors="coerce"
)

if movies["release_year"].dtype == "object":
    numeric_as_text = movies[
        movies["release_year"].astype(str).str.fullmatch(r"\d+")
    ]
else:
    numeric_as_text = pd.DataFrame()

invalid_release_year = movies[
    movies["release_year_num"].isna()
]

out_of_range_year = movies[
    (movies["release_year_num"] < 1900) |
    (movies["release_year_num"] > 2026)
]

activity["watch_date_dt"] = pd.to_datetime(
    activity["watch_date"], errors="coerce"
)

invalid_watch_dates = activity[
    activity["watch_date_dt"].isna()
]

print("Numeric-as-text values :", len(numeric_as_text))
print("Malformed release_year :", len(invalid_release_year))
print("Out-of-range years     :", len(out_of_range_year))
print("Malformed watch_date   :", len(invalid_watch_dates))

print("\n2. INVALID / OUT-OF-RANGE VALUES")

valid_languages = [
    "English",
    "Hindi",
    "Telugu",
    "Tamil",
    "Kannada",
    "Malayalam",
    "Spanish",
    "French",
    "Japanese",
    "Korean"
]

if "language" in movies.columns:
    invalid_language = movies[
        ~movies["language"].isin(valid_languages)
    ]
    print("Invalid language values :", len(invalid_language))
else:
    invalid_language = pd.DataFrame()
    print("Language column not found.")

invalid_completion = activity[
    (activity["completion_percentage"] < 0) |
    (activity["completion_percentage"] > 100)
]

print("Invalid completion_percentage :", len(invalid_completion))

print("\n3. CONTENT TYPE CONSISTENCY")

if "content_type" in movies.columns:

    movies["content_type_clean"] = (
        movies["content_type"]
        .astype(str)
        .str.strip()
        .str.title()
    )

    inconsistent = movies[
        movies["content_type"] != movies["content_type_clean"]
    ]

    print("Unique Categories:")
    print(movies["content_type_clean"].unique())

    print("Inconsistent Categories :", len(inconsistent))

else:
    inconsistent = pd.DataFrame()
    print("content_type column not found.")

print("\n" + "=" * 60)
print("PROFILING REPORT FINDINGS")
print("=" * 60)

if (
    len(numeric_as_text) == 0 and
    len(invalid_release_year) == 0 and
    len(out_of_range_year) == 0 and
    len(invalid_watch_dates) == 0
):
    print("No data type issues detected.")
else:
    print("Data type issues detected.")

if (
    len(invalid_language) == 0 and
    len(invalid_completion) == 0
):
    print("No invalid or out-of-range values detected.")
else:
    print("Invalid or out-of-range values detected.")

if len(inconsistent) == 0:
    print("Content type categories are consistent.")
else:
    print("Content type categories are inconsistent.")

print("\nOverall Status:")
print("Datasets are ready for Exploratory Data Analysis (EDA).")
```
